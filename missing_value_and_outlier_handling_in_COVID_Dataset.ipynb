#Distribution

```python
def plot_per_column_distribution(df, nrows, ncols):
    numerical_columns = df.select_dtypes(include=['float64', 'int64']).columns
    # Calculate the number of subplots needed
    num_plots = len(numerical_columns)
    # Calculate appropriate number of rows and columns for subplots
    nrows = (num_plots + ncols - 1) // ncols
    fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(15, 10))
    for i, column in enumerate(numerical_columns):
        # Handle potential for extra subplots
        if i >= num_plots:
            break
        ax = axes[i // ncols, i % ncols]
        df[column].plot(kind='hist', ax=ax, bins=20)
        ax.axvline(df[column].mean(), color='r', linestyle='--', label='Mean')
        ax.axvline(df[column].median(), color='g', linestyle='-', label='Median')
        ax.set_title(column)
        ax.legend()
    plt.tight_layout()
    plt.show()

# Adjust the number of rows as needed based on the number of numerical columns
plot_per_column_distribution(df, 5, 3)
```

```python
def out_iqr(df , column):
    global lower,upper
    q25, q75 = np.quantile(df[column], 0.25), np.quantile(df[column], 0.75)
    # calculate the IQR
    iqr = q75 - q25
    # calculate the outlier
    cut_off = iqr * 1.5
    # calculate the lower and upper bound value
    lower, upper = q25 - cut_off, q75 + cut_off
    print('The IQR is',iqr)
    print('The lower bound value is', lower)
    print('The upper bound value is', upper)
    # Calculate the number of records below and above lower and above bound value respectively
    df1 = df[df[column] > upper]
    df2 = df[df[column] < lower]
    return print('Total number of outliers are', df1.shape[0]+ df2.shape[0])
```

Using Z-Sscore :

```python
def detect_outliers_zscore(df, column, threshold=3):
    # Calculate mean and standard deviation
    mean = df[column].mean()
    std_dev = df[column].std()
    # Compute Z-score for each data point
    z_scores = (df[column] - mean) / std_dev
    # Identify outliers based on the threshold
    outliers = (np.abs(z_scores) > threshold)
    return outliers
outliers = detect_outliers_zscore(df1, 'INTUBED')
print('Total number of outliers are', outliers.sum())
```

#### Overview
The dataset for analysis and modeling consists of combined data from two sources: a training dataset and a test dataset. The training dataset includes historical data from 2000 to 2019, and the test dataset provides data for the years 2016 to 2020. The dataset is structured to facilitate classification tasks based on socio-economic and environmental indicators.

#### Columns
The combined dataset includes the following columns:

1. **Access to electricity (% of population)**: Percentage of the population with access to electricity.
2. **Agricultural land (% of land area)**: Percentage of land area used for agricultural purposes.
3. **Annual freshwater withdrawals, total (% of internal resources)**: Percentage of total internal freshwater resources withdrawn annually.
4. **Arable land (% of land area)**: Percentage of land area that is arable.
5. **Forest area (% of land area)**: Percentage of land area covered by forests.
6. **Electric power consumption (kWh per capita)**: Average electric power consumption per person measured in kilowatt-hours.
7. **Energy use (kg of oil equivalent per capita)**: Average energy use per person measured in kilograms of oil equivalent.
8. **Renewable electricity output (% of total electricity output)**: Percentage of electricity generated from renewable sources.
9. **Renewable energy consumption (% of total final energy consumption)**: Percentage of total energy consumption derived from renewable sources.
10. **Population growth (annual %)**: Annual growth rate of the population.
11. **GDP per capita (current US$)**: Gross domestic product per person measured in current US dollars.
12. **CO2 emissions (metric tons per capita)**: Average CO2 emissions per person measured in metric tons.

- **Training Dataset**: Covers the years 2000 to 2015.
- **Test Dataset**: Provides data for the years 2016 to 2020.
